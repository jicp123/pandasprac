```python
import pandas as pd
students = pd.read_csv("students_complete.csv")
schools = pd.read_csv("schools_complete.csv")
students = students.set_index("student_name").rename_axis("students", axis="columns")
pd.set_option('display.max_rows', 5)
```

```python

students["avg_score"] = (students["reading_score"] + students["math_score"])/2
students.head()
```

```python
#avg scores of students in each school
students.groupby("school_name").avg_score.mean()
```
